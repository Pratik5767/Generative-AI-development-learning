```python
user_query = "I am an Indian and i live in india"
```

```python
document = "India is a country for the indians and everyone"
```

### Tokenizing / Chunking

```python
query_tokens = user_query.lower().split(' ')
```

```python
query_tokens
```

```python
document_tokens = document.lower().split(' ')
```

```python
document_tokens
```

### Embedding

```python
query_counter = Counter(query_tokens)
```

```python
query_counter
```

```python
document_counter = Counter(document_tokens)
```

```python
document_counter
```

```python
lst = []
for token in query_counter.keys():
    print(query_counter[token])
```

```python
lst = []
for token in query_counter.keys():
    lst.append(query_counter[token])
```

```python
# Sentence vector
lst
```

### Cosine Similarity

```python
mylst = []
for tokens in query_counter.keys() & document_counter.keys():
    # print(tokens)
    mylst.append(query_counter[tokens] * document_counter[tokens])
```

```python
mylst
```

```python
dot_product = sum(mylst)
```

```python
query_magnitude = math.sqrt(sum(query_counter[tokens] ** 2 for tokens in query_counter))
```

```python
query_magnitude
```

```python
document_magnitude = math.sqrt(sum(document_counter[tokens] ** 2 for tokens in document_counter))
```

```python
document_magnitude
```

```python
similarity = (dot_product) / (query_magnitude * document_magnitude)
```
